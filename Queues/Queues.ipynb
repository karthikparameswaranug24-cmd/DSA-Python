## What is a Queue?

#### A linear data structure, that follows the FIFO (first in first out) principle

## Common Operations:

#### 1. Enqueue:- Inserting element at the rear
#### 2. Dequeue:- Removing element from front
#### 3. FrontPeek:- Accessing the front element
#### 4. RearPeek:- Viewing the element at the rear
#### 5. isEmpty:- To check whether or not queue is empty

## Ways of Implementing Queues in Python:

#### 1. Using Lists
#### 2. Using Linked Lists
#### 3. Using the module 'collections', which has 'deque' object and is the most optimized way of implementing queue for DSA/CP purposes
#### 4. Using another module called 'heapq', which is specific for implementing Priority Queues
#### 5. Using 'queue' module, which looks as straightforward as the name suggests. But, is actually used for locking and blocking operations; most suitable for cross-thread communications. Hence, is for very advanced use cases.
#### 6. Module 'asyncio', which is mainly used when working with web applications

## 1. Implementing Queues Using Basic Python Lists:-

```python
# Coming up with our basic initial setup of our queue:
class QueueUsingList:

    # Constructor:
    def __init__(self):

        self.queue = []

    def is_empty(self):

        return len(self.queue) == 0

    def display(self):

        if self.is_empty():

            print('Cannot display.. Queue is empty')

            return  

        print(self.queue)

    def enqueue(self, data):

        self.queue.append(data)

        self.display()

    def dequeue(self):

        if self.is_empty():

            print('Queue is empty')

            return

        return self.queue.pop(0)

    def front_peek(self):

        if self.is_empty():
        
            print('Cannot display.. Queue is empty')

            return 

        return self.queue[0]

    def rear_peek(self):

        if self.is_empty():
        
            print('Cannot display.. Queue is empty')

            return 

        return self.queue[-1]
```

## 2. Implementing Queues Using a Linked List:-

```python
# To implement queue, we need to have two reference pointers--front and rear
# Using these two, we will carry out queue operations.

# To implement Linked List, we need to come up with a node:

class Node:

    # The following is my constructor:

    def __init__(self, data):

        self.data = data

        self.next = None 

    
```

```python
class QueueUsingLL:

    def __init__(self):

        self.front = self.rear = None # Our queue is initially empty


    def is_empty(self):

        return self.front is None  

    def display(self):

        if self.is_empty():

            print('Cannot display... Queue is empty')

            return 


        # But, what if the queue is not empty?
        # To display elements we will start from front and go to the rear

        curr = self.front

        while curr:

            print(curr.data, sep = " ")

            curr = curr.next

        print()

    def enqueue(self, data):

        node = Node(data)

        # Elements are to be inserted from the rear side

        # Case-1:- If the queue is empty:
        if self.is_empty():

            self.front = self.rear = node 

        # Case-2:- The queue is not empty:
        else:

            self.rear.next = node 

            self.rear = node 


    def dequeue(self):

        if self.is_empty():
        
            print('Cannot dequeue... Queue is empty')

            return

        # But, what if the queue is not empty:

        curr = self.front 

        self.front = curr.next

        # In case of a queue of only one element:
        if not self.front:

            self.rear = None 

        return curr.data

    def front_peek(self):

        if self.is_empty():
        
            print('Cannot peek from front... Queue is empty')

            return

        return self.front.data

    def rear_peek(self):

        if self.is_empty():
        
            print('Cannot peek from rear... Queue is empty')

            return

        return self.rear.data
```

```python
from collections import deque
```

## 1. Implement Queue Using Stack

```python
class QueueUsingStack:

    def __init__(self):

        # Initially, our queue is empty
        self.s1 = deque()

        self.s2 = deque()


    def enqueue(self, data):

        self.s1.append(data) # The stack s1 is only for enqueuing purposes

    def dequeue(self):

        if not self.s2:

            while self.s1:

                self.s2.append(self.s1.pop())

        if not self.s2:

            print('Queue is empty.. cannot dequeue')

            return

        return self.s2.pop()

    def peek(self):

        if not self.s2:
            while self.s1:
                self.s2.append(self.s1.pop())

        return -1 if not self.s2 else self.s2[-1]

    def isEmpty(self):

        return not self.s1 and not self.s2 
```

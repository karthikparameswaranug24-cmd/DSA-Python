## Linear search, also called as sequential search, is a fundamental algorithm, which is used to find a specific element within a data structure (array, linked list, etc).

## It operates by sequentially examining each element in the collection until the target element is found or end of collection is reached.

## Working:

#### 1. The search begins at the first element of the data structure.
#### 2. Each element is compared to the target value.
###### (i) If the match is found then, index of the element is returned.
###### (ii) If match is not found then, the algorithm will say that element is not present.

## Characteristics:
#### 1. Easy to understand and implement.
#### 2. It does not require data to be sorted, making it suitable for unordered lists.
#### 3. Time Complexity :- In worst case scenario, O(n) 
#### 4. Space Complexity :- O(1)

```python
# Assuming a non-empty array:
def linear_search(arr, target):

    for index, element in enumerate(arr):

        if element == target:

            return index

    return -1
```

```python
data = [4, 6, 1, 3, -1, 10, 5, 8]

linear_search(data, -1)
```

```python
linear_search(data, 5)
```

```python
linear_search(data, 100)
```

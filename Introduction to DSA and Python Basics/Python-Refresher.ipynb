### List Comprehensions:-

```python
# to print even numbers in the range 1 to 10:
# method-1:
res = []
for i in range(1, 11):
  if i % 2 == 0:
    res.append(i)  
res      
```

```python
# method-2:
list(range(2, 11, 2))
```

```python
# method-3:
# To generate a new sequence out of the existing sequence:
seq = range(1, 11);
[a for a in seq if a % 2 == 0]
```

```python
ages
```

```python
# Another program:
# method-1:
for age in ages:
  if age > 30:
    print("Greater than 30")
  else:
    print("Less than 30")
```

```python
# method-2:
["Greater than 30" if age > 30 else "Less than 30" for age in ages]
```

### Classes and Objects:-

```python
# syntax for class definition:
class Random:   
  def __init__(self):
    pass   
```

```python
# To create a class, which defines a rectangle
class MyRectangle:
  def __init__(self, length, breadth):
    self.length = length
    self.breadth = breadth 

  def get_perimeter(self):
    return 2 * (self.length + self.breadth)
  
  def get_area(self):
    return (self.length * self.breadth)
```

```python
# creating objects of the class
obj1 = MyRectangle(length=5, breadth=6)
```

```python
obj1
```

```python
# to display attributes:
obj1.length
```

```python
obj1.breadth
```

```python
obj1.get_perimeter()
```

```python
obj1.get_area()
```

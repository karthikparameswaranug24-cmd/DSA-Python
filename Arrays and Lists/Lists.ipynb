### 2 D List operations:

creation:

```python
matrix = [
  [1, 2, 3],
  [4, 5, 6],
  [7, 8, 9]
]

print(type(matrix))
```

traversing and displaying:

```python
# directly calling the matrix object:
matrix
```

```python
for row in matrix:
  print(row)
```

```python
def display_matrix(mat):
  for row in mat:
    print(row)
```

```python
display_matrix(matrix)
```

```python
# traversal:
for row in matrix:
  for element in row:
    print(element, end = " ")
  print()
```

```python
# using indices:
m = len(matrix)
n = len(matrix[0])
for i in range(m):
  for j in range(n):
    print(matrix[i][j], end = " ")
  print()  
```

```python
matrix[2][1]
```

updation:

```python
matrix[1][2] = 10
display_matrix(matrix)
```

insertion:

```python
matrix[2].insert(1, 15)
```

```python
display_matrix(matrix)
```

```python
display_matrix(matrix)
```

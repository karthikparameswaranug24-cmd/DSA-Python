## Problem:-
# Reverse the Array

## Problem Statement:

Given an array 'arr' of size 'n',
return an array with all the elements placed in reverse order.

```python
arr = [5, 7, 8, 1, 6, 3]
```

Approach-1:

```python
arr.reverse()
```

```python
arr
```

Approach-2:

```python

arr[::-1]
```

Approach-3:

```python
n = len(arr)
i = 0
j = n-1
while i < j:
  arr[i], arr[j] = arr[j], arr[i]
  i += 1
  j -= 1
print(arr)
```

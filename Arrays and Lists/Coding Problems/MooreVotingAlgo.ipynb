## Given an array of integers of size n, find the element, that occurs more than ⌊n / 2⌋ times.

```python
# If n = 8, then element should appear more than 4 times
# If n = 9, then element should appear more than 4 times
# Given the array:
arr = [2, 2, 3, 3, 1, 2, 2]
# 2 appears for 4 times, which is greater than ⌊n / 2⌋ i.e., 3
# So, 2, over here, is my answer.
```

## Brute-Force Approach:-

```python
def majorityElement(nums):

    n = len(nums)

    for i in range(0, n):

        cnt = 0

        for j in range(0, n):

            if nums[j] == nums[i]:

                cnt += 1

        if cnt > (n // 2):
            return nums[i]

        break

    return -1
```

```python
majorityElement(arr)
```

## Better Approach:-

```python
# Keeping a hashmap with element as the key and count as its value
from typing import List

def majorityElement(nums):
    mp = {}
    n = len(nums)
    for num in nums:
        if num in mp:
            mp[num] += 1
        else:
            mp[num] = 1
    for val, cnt in mp.items():
        if cnt > (n // 2):
            return val
            break
    
    return -1
```

```python
majorityElement(arr)
```

## Optimal Approach:-

```python
# 1. Apply Moore's Voting Algorithm
# 2. Verify that the element, that you got is the majority element

def majorityElement(nums):

    cnt = 0

    el = None
    n = len(nums)
    for i in range(n):
        # This is where I start a check for new section:
        if cnt == 0:
            el = nums[i]
            cnt = 1
        elif nums[i] == el:
            cnt += 1
        else:
            cnt -= 1
    cnt1 = 0
    for i in range(n):
        if nums[i] == el:
            cnt1 += 1
    if cnt1 > (n // 2):
        return el # This element is my answer
    return -1
```

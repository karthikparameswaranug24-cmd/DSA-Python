## Problem:-
# Remove Duplicates from the List

### Problem statement:
Given a list (or sequence) of elements and remove elements, that are repeating more than once

```python
nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
```

Brute-Force Approach:

```python
def removeDuplicates(k):
  n1 = len(nums)
  st = set() 
  for i in range(n1):
    st.add(nums[i])

  temp = list(st)
  n2 = len(temp)
  for i in range(n2):
    nums.insert(i, temp[i])
  return n2
```

```python
n = len(nums)
n
```

```python
removeDuplicates(n)
```

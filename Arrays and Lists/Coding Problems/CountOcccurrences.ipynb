## Problem statement:-

You are given a list 'items' and a target element 'target'. Return the number of times the 'target' element occurs in a list.

```python
# Given the following:
items = ["apple", "banana", "apple", "orange", "banana", "apple"]
target = "apple"
```

```python
def return_count(items, target):
  count = 0
  n = len(items)
  for i in range(n):
    if items[i] == target:
      count += 1
  return count      
```

```python
countElements = return_count(items, target)
countElements
```

## Given an integer array nums, find the subarray with largest sum and return its sum

```python
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
```

## Brute-Force Approach:-

```python
# We will generate all the subarrays
# Whichever subarray gives the maximum sum, that subarray will be my answer

def maxSubarray(nums) -> int:
    maxi = 0

    n = len(nums)

    for i in range(0, n):

        for j in range(0, n):

            sum = 0

            for k in range(i, j+1):

                sum += nums[k]

            maxi = max(sum, maxi)

    return maxi
```

```python
maxSubarray(nums)
```

```python
nums = [1]
maxSubarray(nums)
```

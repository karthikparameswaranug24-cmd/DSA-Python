## 1. Reverse String Using Stack
#### Assume a non-empty string. You have to simply reverse it.

```python
from collections import deque

def reverseStr(string):

    res = '' # Initializing our resultant string with an empty ones

    st = deque()

    for char in string:

        st.append(char)

    # Popping each character from the stack

    while st:

        popped = st.pop()

        res += popped

    return res
```

```python
reverseStr('happy')
```

```python
reverseStr('This is a string')
```

## 2. Check for Balanced Parentheses



```python
def isValidParenthesis(expression) -> bool:
    stack = deque()

    pairs = {
        ')' : '(',
        ']' : '[',
        '}' : '{' 
    }

    for char in expression:

        if char in '({[':

            stack.append(char)

        elif char in ')}]':

            pair = pairs[char]

            if (not stack) or (pair != stack.pop()):
                return False 

    return len(stack) == 0
```

```python
isValidParenthesis('([{([{}])}])')
```

```python
isValidParenthesis('([{([{}])}]))')
```

```python
isValidParenthesis('{([{([{}])}])')
```

```python
isValidParenthesis('(([{([{}[]()])}])')
```

```python
isValidParenthesis('[[}[')
```

## 3. Undo Operation Simulation
#### Given a list of commands, recreate the undo operation.

```python
def simulate_undo(commands):
    st = deque()

    for command in commands:

        if command == 'undo':

            if st:
                st.pop() # Removing from the stack the last letter, that we typed 

            else:
                print("Empty stack.. cannot UNDO")

                return 

        else:

            st.append(command[-1]) # Pushing into the stack, the letter we are required to type

    return ''.join(st) # deque is very similar to a list in terms of functionality 
```

```python
commands = [
    'type A', 
    'type B',
    'undo',
    'type C' 
]
```

```python
res = simulate_undo(commands)

print(f'Result: {res}')
```

```python
commands = [
    'type A', 
    'type B',
    'undo',
    'type C',
    'undo' 
]

res = simulate_undo(commands)

print(f'Result: {res}')
```

## 1. Reverse String Using Stack
#### Assume a non-empty string. You have to simply reverse it.

```python
from collections import deque

def reverseStr(string):

    res = '' # Initializing our resultant string with an empty ones

    st = deque()

    for char in string:

        st.append(char)

    # Popping each character from the stack

    while st:

        popped = st.pop()

        res += popped

    return res
```

```python
reverseStr('happy')
```

```python
reverseStr('This is a string')
```

## 2. Check for Balanced Parentheses



```python
def isValidParenthesis(expression) -> bool:
    stack = deque()

    pairs = {
        ')' : '(',
        ']' : '[',
        '}' : '{' 
    }

    for char in expression:

        if char in '({[':

            stack.append(char)

        elif char in ')}]':

            pair = pairs[char]

            if (not stack) or (pair != stack.pop()):
                return False 

    return len(stack) == 0
```

```python
isValidParenthesis('([{([{}])}])')
```

```python
isValidParenthesis('([{([{}])}]))')
```

```python
isValidParenthesis('{([{([{}])}])')
```

```python
isValidParenthesis('(([{([{}[]()])}])')
```

```python
isValidParenthesis('[[}[')
```

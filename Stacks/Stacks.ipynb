## What is a Stack?
#### 1. A linear data structure, that stores elements in a sequential order.
#### 2. Insertion and deletion happens only on one end called 'top' of the stack.
#### 3. Stacks follow a Last In First Out (LIFO) principle.

## Implementation Using Lists:-

```python
class MyStack:

    def __init__(self):
        self.stack = []

    def get_size(self):

        return len(self.stack)

    def is_empty(self):

        return self.get_size() == 0

    def display(self):

        if self.is_empty():

            print("Empty stack")

            return 

        for ele in self.stack[::-1]:
            print(ele)

        print() # For readability

    # Moving on to our first operation i.e., push:

    def push(self, data):

        self.stack.append(data)

        self.display()

    def pop(self):

        if self.is_empty():

            print("Empty stack")

            return None

        popped = self.stack.pop()

        return popped 

    def peek(self):

        if self.is_empty():

            print("Empty stack")

            return None 

        print("Peek element ", self.stack[-1])
```

```python
# Now, using our stack:
stack = MyStack()
```

## Implementing stack using linked list

```python
# Node of a singly linked list
class Node:
    def __init__(self, data):
        self.data = data 

        self.next = None 
```

```python
# Creating the stack class using linked list:
class StackUsingLL:

    # Making the constructor:
    def __init__(self):

        self.top = None # Stack's initial condition is that of its emptiness

    # Checking for the emptiness of the stack:
    def is_empty(self):

        return self.top is None 

    def display(self):
        if self.is_empty():
            print('Empty stack.. cannot display')

            return 

        curr = self.top

        while curr:

            print(curr.data)

            curr = curr.next

        print()

    def push(self, data):
        node = Node(data)

        node.next = self.top 

        self.top = node 

        self.display()


    def pop(self):

        if self.is_empty():
            print('Empty stack.. cannot pop')

            return None 

        popped = self.top.data

        self.top = self.top.next

        return popped 

    def peek(self):
        if self.is_empty():
            print('Empty stack.. cannot peek')

            return None 

        return self.top.data
```

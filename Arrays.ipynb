# Arrays :-

## Using Array module:

```python
import array
```

```python
# Array creation
arr1 = array.array('i', [10, 20, 30, 40, 50])
print(arr1)
```

Accessing elements:

```python
# accessing elements
arr1[2]
```

```python
arr1[3]
```

Array traversal:

```python
for element in arr1:
  print(element)
```

```python
for element in arr1:
  print(element, end=" ")
```

Insertion:

```python
arr1.insert(3, 35)
```

```python
print(arr1, end=" ")
```

deletion:

```python
arr1.remove(30) # .remove() removes only the first occurrence of the given element in the array
```

```python
print(arr1)
```

updation:

```python
# to update the third element of the array:
arr1[2] = 30
print(arr1)
```

Sorting:

```python
arr2 = array.array('i', [60, 80, 70, 40, 20]) # Implementing another array to implement sorting
print(arr2)
```

```python
array.array('i', sorted(arr2))
```

merging:

```python
arr3 = array.array('i', [90, 100, 110])
print(arr3)
```

```python
arr4 = arr1 + arr3
```

splitting:

```python
ix = int(len(arr4)/2)
```

```python
ix
```

```python
arr4[:ix]
```

```python
arr4[ix:]
```

## Using numpy:

```python
import numpy as np
```

Creation:

```python
np1 = np.array([10, 20, 30, 40, 50])
np2 = np.array([60, 70, 80])
print(np1)
print(np2)
```

accessing elements:

```python
np1[3]
```

```python
np1[2]
```

```python
np1[0]
```

traversal:

```python
for element in np1:
  print(element, end=" ")
```

insertion:

```python
np.insert(np1, 3, 35)
```

deletion:

```python
np.delete(np1, 3)
```

updation:

```python
np1[2] = 35
```

```python
print(np1, end = " ")
```

sorting:

```python
# creating another array to implement sorting
np3 = np.array([5, 3, 7, 6, 2, 9])
print(np3, end=" ")
```

```python
np.sort(np3)
```

merging:

```python
np3 = np.concatenate((np1, np2)) # merges the array, that you will specify
np3
```

splitting:

```python
np.array_split(np3, 2)
```

##### String Formatting:

```python
# Using f-strings:
```

```python
first = 'James'
last = 'Anderson'
age = 25
res = f'My name is {first} {last} and I am {age} years old.'
print(res)
```

```python
res = f'My name is {} {} and I am {} years old.'.format(first, last, age)
print(res)
```

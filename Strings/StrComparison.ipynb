### Case Sensitivity:

```python
# Now let's say we have:
str1 = 'zoo'
str2 = 'zoO'
```

```python
str1 == str2
```

```python
str1 != str2
```

```python
str1 < str2
```

```python
str1 <= str2
```

```python
str1 > str2
```

```python
str1 >= str2
```

```python
ord('o'), ord('O')
```

```python
for ch1, ch2 in zip(str1, str2):
    print(ch1, ord(ch1), '|', ch2, ord(ch2))
```

### Sorting:

```python
str1 = 'abc ABC'
print(str1)
```

```python
# We can use the built-in sorted() function:
sorted(str1)
```

```python
for ch in str1:
    print(ch, ord(ch))
```

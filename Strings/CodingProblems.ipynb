## Problem-1: Count Vowels and Consonants
#### Evidently, given a string, assume that all the characters are lowercase and may include special characters. Count the number of vowels and consonants.

```python
s = ' __hel lo 123$world  '
```

```python
def isVowel(char):
    # Since all chars are of lowercase,
    
    if char == 'a' or char == 'e' or char == 'i' or char == 'o' or char == 'u':
        return True 
        
    return False 

def vowelAndConsonant(s):
    n = len(s)

    vowelCnt = 0

    consoCnt = 0
    

    res = []

    for i in range(n):
        if not s[i].isalpha():
            continue

        else:
            if isVowel(s[i]):

                vowelCnt += 1

            else:
                consoCnt += 1

    res.append([vowelCnt, consoCnt])
    return res
```

```python
str = input("Enter string: ")
print(str)
out = vowelAndConsonant(str)
print(out)
```

## Problem-2: Count Frequency of Characters

#### You have been given a string of alphabets only. Return the count of each character.

```python
# For example:
s = 'banana'  
# Here, result: b = 1, a = 3, n = 2
```

Using dictionary approach

```python
# Approach-1:
def freqOfChars(s):
    freqDict = {}

    n = len(s)

    for i in range(n):
        freqDict[s[i]] = s.count(s[i])

    return freqDict
```

```python
str1 = input("Enter a string: ")
print(str1)

myDict = freqOfChars(str1)
print(myDict)
```

```python
# Approach-2:
def freqOfChars(s):
    counts = dict()

    for char in s:
        counts[char] = counts.get(char, 0) + 1

    return counts
```

```python
str2 = input("Enter a string: ")
print(str2)

countOfChar = freqOfChars(str2)
print(countOfChar)
```

Using built-in collections module

```python
from collections import Counter 
def freqOfChars(s):
    
    return Counter(s)
```

```python
str3 = input("Enter a string: ")
print(str3)
cntOfChar = freqOfChars(str3)
print(cntOfChar)
```

## Problem-3: Palindrome String

#### Check whether a given string is palindrome or not.

```python
# For example
s = 'madam' 
```

```python
def isPalindrome(s):
    cleaned = ""  

    n = len(s)

    for i in range(n):
        if not s[i].isalnum():
            continue
        else:
            cleaned += s[i].lower() 

    new_n = len(cleaned)

    low = 0
    high = new_n - 1

    while low < high:
        if cleaned[low] != cleaned[high]:
            return False
            break 
        else:
            low += 1
            high -= 1

    return True  
        
```

```python
s2 = input("Enter a string: ")
print(s2)
Palindrome = isPalindrome(s2)
print(Palindrome)
```

## Problem-4: Check for Anagrams

#### Return True if:
#### 1. Both strings contain same characters in any order
#### 2. Strings will be of same length and contains lowercase letters

```python
# Approach-1:
def isAnagram(s, t):

    sorted_s = sorted(s)
    sorted_t = sorted(t)

    n = len(sorted_s)
    m = len(sorted_t)

    if n != m:
        return False 
    
    else:
        for i in range(n):
            if sorted_s[i] != sorted_t[i]:
                return False 
                break 

    return True 
```

```python
s = input("Enter first string: ")
t = input("Enter second string: ")

print(s)
print(t)

validAnagram = isAnagram(s, t)

print(validAnagram)
```

```python
# Approach-2:
# Leveraging the fact that there are exactly 26 alphabets in English language
def isAnagram(s, t):
    mapping = [0] * 26

    n = len(s)
    m = len(t) 

    if n != m:
        return False
    
    else:
        for ch in s:
            idx = ord(ch) - 97

            mapping[idx] += 1

        for ch in t:
            index = ord(ch) - 97

            mapping[index] -= 1

    if all(m == 0 for m in mapping):
        return True
    return False 
```

```python
s = input("Enter first string: ")
t = input("Enter second string: ")

print(s)
print(t)

validAnagram = isAnagram(s, t)

print(validAnagram)
```

## What is a Linked List?
#### 1. A linear data structure composed of nodes, connected by pointers (or references)
#### 2. Each node contains a data and pointer to the next node in sequence.
#### 3. Linked Lists do not require contiguous memory and can dynamically change size.

#### Singly Linked List:

#### Creation

```python
# Initializing the Node class
class Node:
    def __init__(self, data): # Constructor
        self.data = data 
        self.next = None 
```

```python
# Creating the class SinglyLinkedList:
class SinglyLinkedList:
    def __init__(self): # Constructor
        self.head = None # No Linked List is empty as of now  

    def get_length(self):
        length = 0
        curr = self.head 
         
        while curr:
            curr = curr.next
            length += 1

        return length 

    def display(self):
        if not self.head:
            print("Empty list")

        else:
            curr = self.head 

            while curr:
                print(curr.data, end=" ")
                curr = curr.next 

            print()

    def insert_at_starting(self, value):
        node = Node(value)

        if self.head:
            node.next = self.head 
            
        self.head = node

        self.display()

    def insert_at_end(self, value):
        node = Node(value)

        # Case-1: when the list is not empty
        if self.head:
            curr = self.head 

            while curr.next:

                curr = curr.next

            curr.next = node 
        
        # Case-2: when the list is empty
        else:

            self.head = node 

        self.display()

    def insert_in_middle(self, pos, value):
        n = self.get_length()

        if (pos < 0) or (pos > n):
            print("Invalid position")

        elif pos == 0:
            self.insert_at_starting(value)

        elif pos == n:
            self.insert_at_end(value)

        else:
            node = Node(value)

            p = self.head
            q = None  

            for _ in range(pos):

                q = p
                p = p.next

            q.next = node 
            node.next = p 

        self.display()

    def search(self, key):

        if not self.head:

            return -1 # List is empty
        
        pos = 0

        curr = self.head 

        while curr:

            if curr.data == key:
                return pos
                break
            
            curr = curr.next
            pos += 1

        return -1

    def delete_at_beginning(self):

        """
        Deletes node at the begginning of Linked List
        """
        if self.head:
            curr = self.head 

            self.head = curr.next 

            curr.next = None 

            self.display()  

        else:
            print("Cannot delete from empty list")

    def delete_at_end(self):

        """
        Deletes node at the end of the Linked List
        """

        if self.head:
            curr = self.head 

            prev = None 

            while curr.next:
                prev = curr 

                curr = curr.next 

            if curr == self.head:
                self.head = None 

            else:
                prev.next = None  

            self.display()

        else:
            print("Cannot delete from empty list")

    def delete_in_middle(self, pos):

        """
        Given any index value, deletes node at the index in the Linked List 
        """

        n = self.get_length()

        if (pos < 0) or (pos >= n):
            print("Invalid index")

        elif pos == 0:
            self.delete_at_beginning()

        elif pos == n-1:
            self.delete_at_end()

        else:
            prev = None  
            curr = self.head 

            for _ in range(pos):
                prev = curr 
                curr = curr.next 

            prev.next = curr.next 
            curr.next = None 

            self.display()
          
```
